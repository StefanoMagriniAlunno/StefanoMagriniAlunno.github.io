### Report1
    
Source: [Perplexity session](https://www.perplexity.ai/search/97da90a5-9ef5-45f2-9ccd-43827a5ddc3f)

Il "Jin-Xin hyperbolic relaxation system in one spatial dimensions" reads:
$$
\begin{cases}
    \partial_t u + \partial_x v = 0 \\
    \partial_t v + a \partial_x u = \frac1\varepsilon\left(f\left(u\right)-v\right)
\end{cases}  \quad a>0, \varepsilon>0
$$
The first equation satisfies the scalar conservation law, the second equation forces the local equilibrium $v=f(u)$.

For the inviscid Burgers' equation, $f(u)=\frac12 u^2$.

The parameter $\varepsilon$ controls the stiffness of the system. A scheme is called Asymptotic-Preserving (AP) if it degenerates to a consistent, stable discretisation of the limiting conservation law as $\varepsilon\to0$, uniformly in $\varepsilon$.

**Chapman-Enskog Expansion**

Applying the Chapman–Enskog expansion to the relaxation system, the first-order correction yields:
$$
    \partial_t u + \partial_x f(u) = \varepsilon \partial_x\left[\left(a-f'\left(u\right)^2\right)\partial_x u\right] + O\left(\varepsilon^2\right)
$$
The sub-characteristic condition $a\geq f'(u)^2$   guarantees positive artificial viscosity and controls dissipation. For Burgers' equation, a standard choice is $a=\max\left|u\right|^2 + \delta$ for a small constant $\delta>0$. This expansion shows that the Jin–Xin system adds an $\varepsilon$-proportional diffusion, which is exactly what smooths out shocks; it is the key idea exploited by the RelaxNN family.

#### Why Standard PINNs Fail

For the Jin-Xin system with small $\varepsilon$, the standard PINN loss is:
$$
    \mathcal{R}^\varepsilon_\textrm{PINN} = \left\|\partial_t u^\textrm{NN} + \partial_x v^\textrm{NN}\right\|^2+\left\|\varepsilon\left(\partial_t v^\textrm{NN}+a\partial_x u^\textrm{NN}\right)-\left(f\left(u^\textrm{NN}\right) - v^\textrm{NN}\right)\right\|+\textrm{BC/IC terms}.
$$
For the Burgers' equation (or the Jin-Xin equilibrium limit when $\varepsilon=0$), the solution develops discontinuities. The strong-form PDE residual $\partial_t u + \partial_x f(u)$ is not well-defined at a shock, so minising its $L^2$ Monte Carlo approximation leads to a stagnant PDE loss and a solution that neither converges to a shock nor satisfies the entropy condition.

#### Strategies

The **first strategy** is Asymptotic-Preserving Neural Networks (APNNs) via Micro-Macro Decomposition. The core idea is define a loss function such that, as $\varepsilon\to0$, the training loss function of the microscopic system converges to the loss function of the correct macroscopic (limit) system.

The **second strategy** is Relaxation Neural Networks (RelaxNN) to Capture Shocks. We know that Jin-Xin relaxation system is a smooth (regularised) solution for any fixed $\varepsilon>0$. The Neural Network approaches the true shock wave with a smooth version. In this way the NN preserves the law described by the equation without unknown error.

The **third strategy** is Weak-Form and Entropy PINNs for Shocks. A wPINN (weak-PINN) is trained with an other kind of loss function:
$$
    \mathcal{L}_\textrm{wPINN} = \sup_{\left\|\phi\right\|_V\leq1}\left|\int_\Omega\left(u^\textrm{NN}\partial_t\phi + f\left(u^\textrm{NN}\right)\partial_x\phi\right)dxdt+\int_\Omega u_0\phi\left(0,\cdot\right)dx\right|^2
$$

#### Comparative Analysis
The methods APNN-like are AP for $\varepsilon\to0$ but are not shock-handling. With an error to the AP, the RelaxNN methods are shock-handling with smooths solutions. The neural network AP-PINNs/Lei (2025) is AP (for continuous solutions) and shock-handling (demonstrated only for shock tube experiment).

#### Open Problems
**AP + Shock:** The most pressing open problem is designing a single framework that is simultaneously AP in $\varepsilon$ and correctly resolves shocks at $\varepsilon=0$. RelaxNN / AP-PINN approach provides AP in a soft sense , but requires tuning $\varepsilon$ and does not recover a crisp discontinuity in the limit. A convergence theorem for discontinuous solutions remains open even for the RelaxNN framework.

**Optimal Selection of the Relaxation Parameters:** For RelaxNN, choosing $\varepsilon$ too large introduces excessive artificial viscosity; too small leads to training instability. A *curriculum strategy* was suggested as a promising direction.

**2 and 3 Dimensional problems:** Most rigorous analysis and error bounds have been established for 1-D systems. Extension to 2-D and 3-D introduces additional challenges: shock surfaces are curves/surfaces, and the spectral bias of DNNs interacts more severely with high-dimensional training domains.

As said in the Report1, the Jix-Xin is a relaxation system parametrized over $\varepsilon$ the limit $\varepsilon\to0$ gives the Burgers' equation.
We see 2 interacting failure modes:
  - Stiffness-Induced gradient pathology: the Neural Tangent Kernel (NTK) norm grows as $O\left(\varepsilon^{-2}\right)$
  - Shock-discontinuity failure: the strong-form $L^2$ PDE residual is structurally inappropriate near shocks, causing loss stagnation and erroneous solutions.

We find many solutions of these problems:
  - Asymptotic preserving (AP) loss reformulations
  - Relaxation-based network architectures
  - Weak/Entropy-based loss formulations
  - Adaptive weighting and NTK-informed balancing
  - Causal/Domain-decomposition training strategies

In Report1, we just talk about some point, so here we focus on the missed informations.

#### NTK Stiffness and Gradient Pathology
In a work of Wang et al. (2022), the authors show that MLP suffer not only from spectral bias, but also from a remarkable discrepancy in convergence rates among the different loss components. It has proved that the Frobenious norm of the NTK matrix grows as $O\left(1/\varepsilon^2\right)$.

#### Residual-Based Adaptive Sampling (RAD/RAR-D)
RAR-D is a technique that redistributes collocation points by sampling proportionally to the current PDE residual magnitude. This technique approaches the loss stagnation.

## Summary

1. Jin-Xin is a model that relaxes Burgers' model over the parameter $\varepsilon$, as $\varepsilon\to0$ the result converges to the solution of Burgers' model.
2. In particular, if we have a shockwave, the Jin-Xin model returns smooth results.
3. There exists a work that study the behaviour of a generic MLP with this kind of problems.
4. The research show us the following solutions:
- **APNN**: The loss of the neural network depend on $\varepsilon$ but when $\varepsilon\to0$ the loss converges to something that represents Burgers' model. 
- **RelaxNN**: We extimate the solution without using the hyperparameter $\varepsilon$. In fact, the loss function includes try to solve directly the Burgers' equation with a soft constraint for the condition $v=f(u)$.

### APNN
The core idea is write the velocity as follows: $v=f(u)+\varepsilon g$ where $g$ is the deviation from the origin. We remember that the Burger's equation require $v=f(u)$ but this equivalence is relaxed in Jin-Xin's model.
The equation of the transport begins:
$$
    \partial_t u + \partial_x f(u) + \varepsilon \partial_x g = 0
$$
The neural network learns $u$ and $g$ with the constraint $\langle g\rangle = 0$.
When $\varepsilon\to0$ we obtain the Burgers' equation.


### RelaxNN
The core idea is to work with $2$ neural networks: $u^\textrm{NN}$ and $v^\textrm{NN}$. The loss function contains the residual for the transport equation: $\partial_t u^\textrm{NN}+\partial_x v^\textrm{NN}$ and a flux loss for the equilibrium $v^\textrm{NN} - f\left(u^\textrm{NN}\right)$. The main result in this framework is the total assence of the hyperparameter $\varepsilon$.
